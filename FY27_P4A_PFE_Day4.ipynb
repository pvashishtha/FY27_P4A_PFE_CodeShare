```python
def outer_function():
    x=10
    def inner_function():
        x=2
        x=x+5
        print(f'X={x}')
    inner_function()
    print(f'X={x}')

outer_function()
```

```python
def outer_function():
    x=10
    def inner_function():
        nonlocal x
        x=x+5
        print(f'X={x}')
    inner_function()
    print(f'X={x}')

outer_function()
```

```python
# Write a code to copy a file of large size from one location to another with a limited size
# of main memory, example 1000 bytes only. Copy the file from source to destination using
# an approach of migrating small size chunks of data at a time.

source_file='source.txt'
dest_file='destination.txt'
ram_size=1000

def copy_file_in_parts(source, destination, size):
    source_handle=open(source)
    dest_handle=open(destination, 'w')

    source_handle.seek(0,2)
    fsize=source_handle.tell()
    print(f'File Size: {fsize}')

    chunks=fsize//size
    source_handle.seek(0)
    import time
    for _ in range(chunks+1):
        print(f'Writing Chunk...')
        data=source_handle.read(size)
        dest_handle.write(data)
        dest_handle.flush()
        time.sleep(0.6)
    source_handle.close()
    dest_handle.close()
    print(f'File Copy Successful!')

copy_file_in_parts(source_file, dest_file, ram_size)
```

```python
# Write image binary file to another location using small size ram, chunk wise transfer.

def copy_file_in_parts(source, destination, size):
    source_handle=open(source, 'rb')
    dest_handle=open(destination, 'wb')

    source_handle.seek(0,2)
    fsize=source_handle.tell()
    print(f'File Size: {fsize}')

    chunks=fsize//size
    source_handle.seek(0)
    import time
    for _ in range(chunks+1):
        print(f'Writing Chunk...')
        data=source_handle.read(size)
        dest_handle.write(data)
        dest_handle.flush()
        time.sleep(0.6)
    source_handle.close()
    dest_handle.close()
    print(f'File Copy Successful!')

copy_file_in_parts('panda.jpg', 'panda_copy.jpg', 50000)
```

```python
# Write a function code captcha(N) that generates capthca of N characters that includes upper and lower
# alphabet, numbers, special chars like [!@#$%]

def captcha(N):
    alpha_lower=[chr(x) for x in range(97,123)]
    alpha_upper=[chr(x) for x in range(65,91)]
    nums=[x for x in range(10)]
    sp='!@#$%'
    chars=[alpha_lower, alpha_upper, nums, sp]
    captcha=''
    for _ in range(N):
        captcha+=str(random.choice(random.choice(chars)))
    return captcha

captcha(8)
```

```python
# Write a function code to generate password for user using characters from name and numbers with special chars

def pwd_generator(name):
    name_list=[x for x in name]
    sp='!@#$%&'
    num=[x for x in range(10)]

    chars=[num, sp, name_list]
    pwd=''
    for _ in range(len(name)+3):
        pwd+=str(random.choice(random.choice(chars)))
    return pwd

pwd_generator('prateek')
```
