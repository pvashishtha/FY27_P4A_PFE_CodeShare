```python
# Write a code in python to accept a string from user, and find the length. If length is odd, print the string
# with alternate upper and lower case letters starting from lower case. If length is even, start from upper case
# and print string letters in alternate upper and lower case.
# Example: Input=> python => length=6 [even] Output=> PyThOn
# Example: Input=> hello => length=5 [0dd] Output=> hElLo
st=input()
length=len(st)
if length%2==0:
    for i in range(length):
        if i%2==0:
            print(st[i].upper(), end='')
        else:
            print(st[i].lower(), end='')
else:
    for i in range(length):
        if i%2==0:
            print(st[i].lower(), end='')
        else:
            print(st[i].upper(), end='')
```

```python
# Write a code to accept a comma separated int values from user and place these values in a list. Then separate
# the odd and even values into 2 lists. Finally merge the list such that odd values comes before all even, while 
# preserving their original order of occurence.
values=input().split(',')
li=[]
odd=[]
even=[]
for el in values:
    li.append(int(el))
for val in li:
    if val%2==0:
        even.append(val)
    else:
        odd.append(val)
print(li, odd, even, sep='\n')
```

```python
# Write a code to accept a comma separated int values from user and place these values in a list. Then process
# the list to identify all prime numbers in list and print them as separate list. If no prime, then print INVALID.
values=input().split(',')
li=[]
prime=[]
for el in values:
    li.append(int(el))
for val in li:
    for i in range(2,val//2+1):
        if val%i==0:
            break
    else:
        prime.append(val)
print(li)
print(prime)
```

```python
# Write a code to accept user name and salary, identify the salary grade from match case and display all info
name=input('Enter Name: ')
salary=int(input('Enter Salary: '))
grade=''
match salary:
    case 25000:
        grade='A'
    case 20000:
        grade='B'
    case 15000:
        grade='C'
    case _:
        grade='Unknown'

print(f'{name} is getting {salary} salary under grade {grade}')
```

```python
# Write a code that evaluates customer discount from the type of customer as Gold, Silver or Platinum.
# Gold gets 10%, Silver gets 5% and Platinum gets 15% discount. Based on type of customer from dictionary
# like {'type':'gold'} determine the dicount and return

cust_type={'type':'platinum', 'years':4}
discount=0

match cust_type:
    case {'type':'platinum', 'years':x} if x>2 :
        discount=15
    case {'type':'gold'}:
        discount=10
    case {'type':'silver'}:
        discount=5
    case _:
        discount=2

print(discount)
        
```

```python
# Write a code that processes the list of integers and identify if the list has alternating values.
# A list has alternating values if values increses or decreses aletrnatingly from element to element
# Example: [2,5,3,7,6,9,4] => alternating
# Example: [4,8,5,7,9,1,4] => not alternating

#li=[2,5,3,7,6,9,4]
li=[4,8,5,7,9,1,4]

for i in range(1, len(li)-1):
    if li[i-1]>li[i]>li[i+1] or li[i-1]<li[i]<li[i+1]:
        print('not alternating')
        break
else:
    print('aletrnating')
```
