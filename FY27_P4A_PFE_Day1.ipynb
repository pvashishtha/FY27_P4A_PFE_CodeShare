```python
# There is a machine that is used to cut cloth pieces. It cuts 3 mtr in first second, 2 mtr in second second and 
# 1 mtr in third second. Then it repeats the pattern of cutting indefinitely. If the machine runs for T seconds
# how many mtrs of cloth it will cut?
T=int(input('Enter num of seconds: '))
# 1-3, 2-2, 3-1 => 3 sec = 6 mtr
# T%3 # 0, 1, 2
cut_length=0
full_cycles=T//3
rem=T%3
cut_length=6*full_cycles
if rem==1:
    cut_length+=3
elif rem==2:
    cut_length+=5
else:
    pass

print('Total Length Cut= ', cut_length, 'mtrs')
```

```python
# Write a python program to find whether the given 3-digit number is AMSTRONG number or not.
# A 3-digit amstrong number is one whose sum of cube of its digits is same as number itself.
# Example: 371 => 3**3 + 7**3 + 1**3 => 27 + 343 + 1 => 371
# If number is 3 digit amstrong number print AMSTRONG else print NOT AMSTRONG
# If number is not 3 digit number, then print INVALID
num=int(input())
temp=num
print('Number: ', num)
if 100<=num<=999:
    amstrong_sum=0
    while num>0:
        amstrong_sum+=(num%10)**3
        num//=10
    print('Amstrong Sum: ',amstrong_sum)
    if amstrong_sum==temp:
        print('AMSTRONG')
    else: print('NOT AMSTRONG')
else:
    print('INVALID')
```

```python
# Write a code to find if given number is palindrome or not
num=int(input())
rev=''
temp=num
while temp>0:
    digit=temp%10
    rev+=str(digit)
    temp//=10
if int(rev)==num:
    print('palindrome')
else:
    print('not palindrome')
```

```python
# Write a code to accept a password string and validate for following rules:
# 1. It must be minimum 8 character and maximum 32 characters (len)
# 2. It should contain atleast 1 uppercase and 1 lower case letter (isupper, islower)
# 3. It must have at least 1 numeric digit (isdigit)
# 4. should not contain special characters (isalnum)

pwd=input()
length=len(pwd)
lowercheck=0
uppercheck=0
digitcheck=0

for char in pwd:
    if char.isupper():
        uppercheck+=1
    if char.islower():
        lowercheck+=1
    if char.isdigit():
        digitcheck+=1
if uppercheck and lowercheck and digitcheck and pwd.isalnum() and (8<=length<=32):
    print('Valid Password')
else:
    print('Invalid Password')
```
