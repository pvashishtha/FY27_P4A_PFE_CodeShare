```python
# Write function code to find factorial of a number
def factorial(num):
    fact=1
    if num<=1:
        return fact
    else:
        for i in range(2,num+1):
            fact*=i
    return fact
factorial(5)
```

```python
# Write a function code to check if number is prime or ot, return True or False
def is_prime(num):
    for i in range(2,num//2+1):
        if num%i==0:
            return False
    else:
        return True
    return False
is_prime(13)
```

```python
# Write a function code to anomaly value in the given list as argument. Function code must find the value that
# does not follow pattern of continuously increasing value in a list. a list must have continuously incresing values
# any value that is out of order should be returned. If any such value is not found, return 'OK' else 
# stop the function processing and return the anomaly value.

def anomaly_detector(data_list):
    for i in range(1, len(data_list)):
        if data_list[i-1]>data_list[i]:
            return data_list[i]
    else:
        return 'OK'

anomaly_detector([1,2,3,4,2,6,7,8,9])
```

```python
# Write a function code to accept 2 operand argument and 1 operator argument, then perform the mathematical
# operation with operands based on operator passed. Hint: use match case.

def calculate(op1, op2, operator):
    match operator:
        case '+':
            return op1+op2
        case '-':
            return op1-op2
        case '*':
            return op1*op2
        case '/':
            if op2==0:
                return 0
            return op1/op2
        case '//':
            if op2==0:
                return 0
            return op1//op2
        case '**':
            return op1**op2
        case _:
            return 0

calculate(10,30,'+')
```

```python
# Write a code with recursive function to generate fibonacci series of N terms
def fibonacci(n):
    if n<=0:
        return 0
    if n==1:
        return 1
    return fibonacci(n-1) + fibonacci(n-2)

for i in range(10):
    print(fibonacci(i), end=' ')
```

```python
# Write a function code that accepts a nested list, and returns a flatten list as a result.
# Example: input=> [1, [2,3], [[4,5], 6]]  output=> [1,2,3,4,5,6]
def flatten(nested_list):
    flat=[]
    for item in nested_list:
        if isinstance(item, list):
            flat.extend(flatten(item)) # recursive call
        else:
            flat.append(item)
    return flat

flatten([1, [2,3], [[4,5], 6]])
```

```python
# Write a code to create a countdown timer using recursion.
def countdown(n):
    if n<=0:
        return 0
    import time
    time.sleep(1)
    print(n)
    countdown(n-1)

countdown(5)
```

```python
# Write recursive function for binary search
def binary_search(arr, start_index, end_index, target):
    if start_index>end_index:
        return -1
    mid=(start_index+end_index)//2
    if target==arr[mid]:
        return mid
    elif target<arr[mid]:
        return binary_search(arr, start_index, mid-1, target)
    elif target>arr[mid]:
        return binary_search(arr, mid+1, end_index, target)

li=[1,4,12,14,45,56,77,79,89,99]
binary_search(li,0,len(li),45)
```

```python
# Decorators
# Special type of function that adds the extra capabilities to user defined fucntions
```

```python
def outer(func):
    def inner():
        print('Inner Starts')
        func()
        print('Inner Ends')
    return inner
```

```python
@outer
def greet():
    print('Good Evening!')
```

```python
greet()
```

```python
# Example for Decorator
def timer(func):
    def inner():
        import time
        start=time.time()
        func()
        end=time.time()
        print(f'{func.__name__} took {round(end-start, 4)} seconds to execute')
    return inner
```

```python
@timer
def greet():
    print('Hello from Greet!')
```

```python
greet()
```

```python
@timer
def calculate_size():
    l1=[x for x in range(10000)]
    l2=[x for x in range(10000)]
    l3=[]
    for x in l1:
        for y in l2:
            l3.append(x*y)
    print(f'length of L3 is {len(l3)}')
```

```python
calculate_size()
```

```python
# Create a Python function that uses a lambda function to sort a list of dictionaries by a specific key.
li=[{'name':'prateek', 'age':40},
    {'name':'aditi', 'age':35},
    {'name':'sumit', 'age':32},
    {'name':'rohan', 'age':39},
    {'name':'deepak', 'age':43}]

def sort_by_key(li, key, reverse=False):
    return sorted(li, key=lambda item:item[key], reverse=reverse)

print(sort_by_key(li, 'name'))
print(sort_by_key(li, 'age', True))
```
